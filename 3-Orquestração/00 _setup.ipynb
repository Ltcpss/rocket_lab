#00 - Setup do Ambiente de Dados
**Projeto:** CineData Analytics, Atividade 2
**Objetivo:** preparar toda a infraestrutura do Data Lakehouse antes da ingestão dos dados

## O que o notebook faz:
1. Cria catálogo exclusivo para a atividade
2. Cria schemas de cada camada:`raw`, `bronze`, `silver` e `gold`
3. Cria o Volume `raw.inputs`, que armazena os CSVs brutos.

## Decisões de Projeto
- **Catálogo próprio:** mantém os nomes `bronze.tb_*`, `silver.tb_*` e `gold.*` exigidos no enunciado sem colisão com atividades futuras.

## Passo manual
O upload dos 5 CSVs é feito pela interface (**Catalog - raw - inputs - Upload to this volume**)

```python
# Criação do catálogo
spark.sql("CREATE CATALOG IF NOT EXISTS atividade_2_cinedata")
display(spark.sql("SHOW CATALOGS"))
```

```python
# Criação dos schemas
CATALOGO = "atividade_2_cinedata"

spark.sql(f"USE CATALOG {CATALOGO}")

for schema in["raw","bronze","silver","gold"]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema}")

display(spark.sql("SHOW SCHEMAS"))
```

```python
# Criação do volume
spark.sql(f"CREATE VOLUME IF NOT EXISTS {CATALOGO}.raw.inputs")
```
